```python
%%sql
CREATE TABLE departments (
    dept_id INT PRIMARY KEY,
    dept_name VARCHAR(50),
    location VARCHAR(50)
);

CREATE TABLE employees (
    emp_id INT PRIMARY KEY,
    emp_name VARCHAR(50),
    salary INT,
    dept_id INT,
    manager_id INT,
    hire_date DATE
);

CREATE TABLE orders (
    order_id INT PRIMARY KEY,
    emp_id INT,
    amount INT,
    order_date DATE
);
```

```python
"""Find employees whose salary is greater than the average salary"""
```

```python
%%sql
select
emp_name,
salary
from employees 
where salary > ( select 
AVG(salary) As avgsalary 
from employees
)
```

```python
"""Find the employee with the highest salary.
"""
```

```python
%%sql
select 
emp_name,
salary
from employees 
where salary = (select
max(salary) AS maxsalary
from employees)
```

```python
"""Find employees who earn less than the lowest salary in department 10"""
```

```python
%%sql
select 
emp_name,
salary
from employees 
where salary = (select
min(salary) AS minsalary
from employees)
```

```python
""". Show each employee's salary and the difference from the company average salary."""
```

```python
%%sql
select
e.emp_name,
e.salary ,
    (select avg(m.salary)
    from employees m )As avgsal,
    
    (select e.salary - avg(m.salary)
    from employees m ) As diffsal
    
from employees e ;

```

```python
"""Find the department with the highest average salary."""
```

```python
%%sql
SELECT d.dept_name
FROM departments d
WHERE d.dept_id = (
    SELECT e.dept_id
    FROM employees e
    GROUP BY e.dept_id
    ORDER BY AVG(e.salary) DESC
    LIMIT 1
);
```

```python
%%sql
SELECT
    d.dept_name,
    (
        SELECT AVG(e.salary)
        FROM employees e
        WHERE e.dept_id = d.dept_id
    ) AS avgsal
FROM departments d
ORDER BY avgsal DESC
LIMIT 1;
```
