```python
%%sql
CREATE TABLE emp (
    employee_id INT PRIMARY KEY,
    employee_name VARCHAR(50),
    department_id INT,
    salary INT,
    manager_id INT
);
```

```python
%%sql
CREATE TABLE dep (
    department_id INT PRIMARY KEY,
    department_name VARCHAR(50)
);
```

```python
%%sql
select
e.employee_name,
e.salary,
d.department_name
from emp e
join dep d on e.department_id = d.department_id;
```

```python
"""Find employees who work in the HR department."""
```

```python
%%sql
select 
e.employee_name	,
d.department_name
from emp e
join dep d on e.department_id = d.department_id
where d.department_name = "HR";
```

```python
"""Display employees whose salary is greater than 35,000, along with their department names."""
```

```python
%%sql
select 
e.employee_name,
e.salary,
d.department_name
from emp e
join dep d on e.department_id = d.department_id
where salary > 35000;
```

```python
"""Display all employees and their department names, including employees without a matching department."""
```

```python
%%sql
select
e.employee_name,
d.department_name
from emp e
left join dep d on e.department_id = d.department_id ;
```
